```python
import requests
import pandas as pd

def check_arxiv_html(arxiv_id):
    url = f"https://arxiv.org/html/{arxiv_id}"

    try:
        response = requests.get(url, timeout=15)

        if response.status_code == 200:
            return "Available"
        else:
            return f"Not Available ({response.status_code})"

    except Exception as e:
        return "Error"

metadata_df["html_status"] = metadata_df["arxiv_id"].apply(check_arxiv_html)

metadata_df[["paper_id", "arxiv_id", "html_status"]]
```

```python
import requests
import time

RAW_DIR = os.path.join(PROJECT_DIR, "data/raw")

download_results = []

for _, row in metadata_df.iterrows():

    paper_id = row["paper_id"]
    arxiv_id = row["arxiv_id"]

    url = f"https://arxiv.org/html/{arxiv_id}"
    output_file = os.path.join(
        RAW_DIR,
        f"{paper_id}_{arxiv_id}.html"
    )

    try:
        response = requests.get(url, timeout=30)

        if response.status_code == 200:
            with open(output_file, "w", encoding="utf-8") as f:
                f.write(response.text)

            download_results.append({
                "paper_id": paper_id,
                "arxiv_id": arxiv_id,
                "status": "Downloaded",
                "file": output_file,
                "size_kb": round(len(response.content) / 1024, 2)
            })

        else:
            download_results.append({
                "paper_id": paper_id,
                "arxiv_id": arxiv_id,
                "status": f"Failed ({response.status_code})",
                "file": "",
                "size_kb": 0
            })

    except Exception as e:
        download_results.append({
            "paper_id": paper_id,
            "arxiv_id": arxiv_id,
            "status": f"Error: {str(e)}",
            "file": "",
            "size_kb": 0
        })

    time.sleep(1)

download_df = pd.DataFrame(download_results)

download_df
```

```python
from bs4 import BeautifulSoup
import os
import pandas as pd


def extract_sections_from_html(html_file, paper_id):
    """
    Extract headings and their associated text from an arXiv HTML paper.
    """

    with open(html_file, "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "html.parser")

    records = []

    headings = soup.find_all(["h1", "h2", "h3", "h4"])

    for heading in headings:

        section_title = heading.get_text(" ", strip=True)

        if not section_title:
            continue

        # Collect text until the next heading
        section_text = []

        for element in heading.find_all_next():

            if element.name in ["h1", "h2", "h3", "h4"]:
                break

            if element.name in ["p", "li"]:
                text = element.get_text(" ", strip=True)

                if text:
                    section_text.append(text)

        records.append({
            "paper_id": paper_id,
            "heading_level": heading.name,
            "section_title": section_title,
            "text": " ".join(section_text)
        })

    return pd.DataFrame(records)
```

```python
from bs4 import BeautifulSoup
import pandas as pd


def extract_sections_from_html(html_file, paper_id):
    """
    Extract structured headings and paragraph/list text
    from an arXiv HTML paper.
    """

    with open(html_file, "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "html.parser")

    records = []

    headings = soup.find_all(["h1", "h2", "h3", "h4"])

    for heading in headings:

        section_title = heading.get_text(" ", strip=True)

        if not section_title:
            continue

        section_text = []

        # Start from the elements immediately following this heading
        for element in heading.find_all_next():

            # Stop when another heading is reached
            if element.name in ["h1", "h2", "h3", "h4"]:
                break

            # Extract paragraph text
            if element.name == "p":
                text = element.get_text(" ", strip=True)

                if text:
                    section_text.append(text)

            # Extract list items
            elif element.name == "li":
                text = element.get_text(" ", strip=True)

                if text:
                    section_text.append(text)

        records.append({
            "paper_id": paper_id,
            "heading_level": heading.name,
            "section_title": section_title,
            "text": " ".join(section_text)
        })

    return pd.DataFrame(records)
```

```python
from bs4 import BeautifulSoup
import pandas as pd


def extract_sections_from_html(html_file, paper_id):
    """
    Extract h2-h4 sections and their associated paragraph/list text
    from an arXiv Semantic HTML paper.
    """

    with open(html_file, "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "html.parser")

    records = []

    # h1 is the paper title, so section extraction starts from h2
    headings = soup.find_all(["h2", "h3", "h4"])

    for heading in headings:

        section_title = heading.get_text(" ", strip=True)

        if not section_title:
            continue

        section_text = []

        for element in heading.find_all_next():

            # Stop at the next section heading
            if element.name in ["h2", "h3", "h4"]:
                break

            if element.name == "p":
                text = element.get_text(" ", strip=True)

                if text:
                    section_text.append(text)

            elif element.name == "li":
                text = element.get_text(" ", strip=True)

                if text:
                    section_text.append(text)

        records.append({
            "paper_id": paper_id,
            "heading_level": heading.name,
            "section_title": section_title,
            "text": " ".join(section_text)
        })

    return pd.DataFrame(records)
```

```python
import re

def clean_text(text):
    """
    Clean extracted research-paper text.

    Removes:
    - newline and tab characters
    - invisible Unicode characters
    - numeric in-text citation markers such as [47], [12, 18], [1-3]

    Preserves:
    - scientific numbers
    - percentages
    - model names
    - equations/textual notation
    - technical terminology
    """

    if not isinstance(text, str):
        return ""

    # Replace newline and tab characters with spaces
    text = text.replace("\n", " ")
    text = text.replace("\t", " ")

    # Remove invisible Unicode characters
    text = re.sub(r"[\u200b\u200c\u200d\ufeff]", "", text)

    # Remove numeric citation markers
    # Examples: [47], [12, 18], [1-3], [1, 2, 5-7]
    text = re.sub(
        r"\[\s*\d+(?:\s*[-–]\s*\d+)?(?:\s*,\s*\d+(?:\s*[-–]\s*\d+)?)*\s*\]",
        "",
        text
    )

    # Normalize multiple spaces
    text = re.sub(r"\s+", " ", text)

    # Remove spaces before punctuation
    text = re.sub(r"\s+([,.!?;:])", r"\1", text)

    return text.strip()
```

```python
all_processed = []

for _, row in metadata_df.iterrows():
    paper_id = row["paper_id"]
    arxiv_id = row["arxiv_id"]

    # P01 will be handled separately
    if paper_id == "P01":
        continue

    html_file = os.path.join(
        RAW_DIR,
        f"{paper_id}_{arxiv_id}.html"
    )

    print(f"Processing {paper_id}...")

    # Extract sections from HTML
    sections = extract_sections_from_html(
        html_file,
        paper_id
    )

    # Apply NEW cleaning function
    sections["text"] = sections["text"].apply(clean_text)

    # Remove empty text rows
    sections = sections[
        sections["text"].str.strip() != ""
    ].copy()

    # Reset index
    sections.reset_index(drop=True, inplace=True)

    # Save processed paper
    output_path = os.path.join(
        PROCESSED_DIR,
        f"{paper_id}_sections_clean.csv"
    )

    sections.to_csv(
        output_path,
        index=False,
        encoding="utf-8"
    )

    all_processed.append(sections)

    print(
        f"{paper_id} completed → "
        f"{len(sections)} sections"
    )

print("\nP02–P10 processing completed.")
```
